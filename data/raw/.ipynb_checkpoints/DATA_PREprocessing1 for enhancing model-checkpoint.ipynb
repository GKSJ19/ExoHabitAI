**Handling Missing Value**

```python
# Count missing values per row
missing_count = df2[selected_columns].isnull().sum(axis=1)

# Try different thresholds
for t in [3, 4, 5, 6]:
    count = (missing_count >= t).sum()
    print(f"Rows with ≥ {t} missing values: {count}")
```

```python
# Count missing values per row (only numerical core features)
missing_count = df2[selected_columns].isnull().sum(axis=1)

# Final threshold: drop rows with 6 or more missing values
threshold = 6

# Rows before applying threshold
rows_before = len(df2)

# Apply threshold
df2 = df2[missing_count < threshold]

# Rows after applying threshold
rows_after = len(df2)

print(f"Rows before thresholding: {rows_before}")
print(f"Rows after thresholding: {rows_after}")
```

```python
# ---------------------------------------------
# Handling Missing Data 
# ---------------------------------------------

# 1. Planetary physical features → Median imputation
planetary_features = [
    "pl_rade",
    "pl_bmasse",
    "pl_orbper",
    "pl_orbsmax",
    "pl_eqt",
    "pl_dens"
]

for col in planetary_features:
    df2[col] = df2[col].fillna(df2[col].median())

# 2. Star temperature → Median imputation
df2["st_teff"] = df2["st_teff"].fillna(df2["st_teff"].median())

#3.Impute remaining stellar features using median
df2["st_lum"] = df2["st_lum"].fillna(df2["st_lum"].median())
df2["st_met"] = df2["st_met"].fillna(df2["st_met"].median())
```

```python
df2.isnull().sum()
```

**Outlier Detection**

```python
# ---------------------------------------------
# Outlier Detection & Handling
# ---------------------------------------------

# 1. REMOVE physically impossible values
# (These values cannot exist in reality)

df2 = df2[
    (df2["pl_rade"] > 0) &                      # Planet radius must be positive
    (df2["pl_bmasse"] > 0) &                    # Planet mass must be positive
    (df2["pl_orbper"] > 0) &                    # Orbital period must be positive
    (df2["pl_orbsmax"] > 0) &                   # Semi-major axis must be positive
    (df2["pl_eqt"] >= 0) &                      # Temperature in Kelvin cannot be negative
    (df2["st_teff"] >= 0)                       # Star temperature in Kelvin cannot be negative
]

# 2. CAP realistic extreme values using IQR
# (Extreme but physically possible values)

numerical_features = [
    "pl_rade", "pl_bmasse", "pl_orbper",
    "pl_orbsmax", "pl_eqt", "pl_dens",
    "st_teff", "st_lum", "st_met"
]

for col in numerical_features:
    Q1 = df2[col].quantile(0.25)
    Q3 = df2[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap values instead of removing them
    df2[col] = df2[col].clip(lower_bound, upper_bound)
```

**Feature Engineering:**

Habitability Score Index


```python
# Habitability Score Index (logic only – reference values passed later)

def habitability_score(df2, dist_ref, lum_ref):
    temp_score = 1 - abs(df2["pl_eqt"] - 288) / (350 - 200)
    radius_score = 1 - abs(df2["pl_rade"] - 1)
    distance_score = 1 / (1 + abs(df2["pl_orbsmax"] - dist_ref))
    luminosity_score = 1 / (1 + abs(df2["st_lum"] - lum_ref))

    score = (temp_score + radius_score + distance_score + luminosity_score) / 4
    return score.clip(0, 1)
```

Stellar Compatibility Index

```python
# Stellar Compatibility Index (logic only)

def stellar_compatibility(df2):
    temp_score = 1 / (1 + abs(df2["st_teff"] - 5778))   # Sun-like temperature
    lum_score = 1 / (1 + abs(df2["st_lum"]))            # Stable luminosity

    score = (temp_score + lum_score) / 2
    return score.clip(0, 1)
```

Orbital Stability Factor

```python
# Orbital Stability Factor (logic only)

def orbital_stability(df2, period_ref, dist_ref):
    period_score = 1 / (1 + abs(df2["pl_orbper"] - period_ref))
    distance_score = 1 / (1 + abs(df2["pl_orbsmax"] - dist_ref))

    score = (period_score + distance_score) / 2
    return score.clip(0, 1)
```

Categorical Encoding

```python
# Create star_type from stellar temperature
def get_star_type(teff):
    if teff < 3700:
        return "M"
    elif teff < 5200:
        return "K"
    elif teff < 6000:
        return "G"
    elif teff < 7500:
        return "F"
    else:
        return "A"

df2["star_type"] = df2["st_teff"].apply(get_star_type)

# One-Hot Encoding
df2 = pd.get_dummies(df2, columns=["star_type"], prefix="star")
```
