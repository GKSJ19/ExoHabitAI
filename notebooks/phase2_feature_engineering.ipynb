# Phase 2: Feature Engineering

Objective:
To create additional astrophysical features such as planet mass,
planet density, and stellar luminosity from the cleaned dataset
as described in the project PDF.


```python
#STEP 0.3: Import Libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

```python
#STEP 0.4: Load Preprocessed Data
df_clean = pd.read_csv("../data/processed/preprocessed.csv")
df_clean.shape
```

PHASE 2 — STEP 1: PLANET MASS

⚠️ Very important:
Your Phase 1 cleaned file does NOT contain planet mass.
So in Phase 2, we bring planet mass from the raw dataset, exactly as allowed by the PDF

```python
#STEP 1.1: Load RAW data (only for mass)
df_raw = pd.read_csv(
    "../data/raw/exoplanet_data.csv",
    comment="#",
    engine="python",
    on_bad_lines="skip"
)

df_raw.shape
```

```python
#STEP 1.2: Check available planet mass columns
[col for col in df_raw.columns if 'mass' in col.lower()]
```

```python
#PHASE 2 — STEP 2: ADD PLANET MASS TO CLEAN DATA
#STEP 2.1: Extract planet mass from raw data
df_mass = df_raw[['pl_name', 'pl_bmasse']]
df_mass.head()
```

```python
#STEP 2.2: Merge planet mass into cleaned data
df_clean = df_clean.merge(df_mass, on='pl_name', how='left')
df_clean.shape
```

```python
#STEP 2.3: Check how many planets have mass
df_clean['pl_bmasse'].isnull().sum()
```

```python
#STEP 3.1: Create planet density column
df_clean['pl_density'] = np.where(
    df_clean['pl_bmasse'].notnull(),
    df_clean['pl_bmasse'] / (df_clean['pl_rade'] ** 3),
    np.nan
)
```

```python
#STEP 3.2: Check density column
df_clean['pl_density'].describe()
```

```python
#STEP 3.3: Check missing density count (expected)
df_clean['pl_density'].isnull().sum()
```

```python
#PHASE 2 — STEP 4: STELLAR (STAR) LUMINOSITY
#STEP 4.1: Create Stellar Luminosity Feature
df_clean['st_luminosity'] = (df_clean['st_rad'] ** 2) * ((df_clean['st_teff'] / 5778) ** 4)
```

```python
#STEP 4.2: Validate Stellar Luminosity
df_clean['st_luminosity'].describe()
```

```python
#STEP 4.3: Final Column Check
df_clean.columns
```

```python
#FINAL STEP: SAVE PHASE 2 OUTPUT
df_clean.to_csv("../data/processed/preprocessed_phase2.csv", index=False)
print("Phase 2 feature engineering completed and saved.")
```
