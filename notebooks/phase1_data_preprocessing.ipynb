# Phase 1: Data Cleaning and Preprocessing

Objective:
To clean and preprocess the raw exoplanet dataset by handling missing values,
removing duplicates and invalid records, and preparing a clean dataset
for further analysis as described in the preprocessing document.


```python
#STEP 2: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
#STEP 3: Load Raw Dataset
df = pd.read_csv(
    "../data/raw/exoplanet_data.csv",
    comment="#",
    engine="python",
    on_bad_lines="skip"
)

df.shape
```

```python
#View first few rows
df.head()
```

```python
#Check structure and data types
df.info()
```

```python
#Summary statistics
df.describe()
```

```python
#STEP 5: Check Missing Values & Duplicates
df.isnull().sum().sort_values(ascending=False).head(20)
```

```python
df.duplicated().sum()
```

```python
#STEP 6: Feature Selection
selected_cols = [
    'pl_name',          # Planet name
    'hostname',         # Star name

    # Planet features
    'pl_rade',          # Planet radius
    'pl_orbper',        # Orbital period
    'pl_orbsmax',       # Semi-major axis
    'pl_eqt',           # Equilibrium temperature

    # Star features
    'st_teff',          # Star temperature
    'st_rad',           # Star radius
    'st_mass',          # Star mass
    'st_met',           # Star metallicity

    # Metadata
    'disc_year',
    'discoverymethod',
    'default_flag'
]

df_clean = df[selected_cols].copy()
df_clean.shape
```

```python
#STEP 6.2: Verify Selected Columns
df_clean.head()
```

```python
df_clean.info()
```

```python
#STEP 7: Handling Missing Values
#STEP 7.1: Check Missing Values in df_clean
df_clean.isnull().sum()
```

```python
#A) Numeric columns → Median
num_cols = df_clean.select_dtypes(include=['int64', 'float64']).columns
df_clean[num_cols] = df_clean[num_cols].fillna(df_clean[num_cols].median())
```

```python
#B) Categorical columns → Mode
cat_cols = df_clean.select_dtypes(include=['object']).columns
for col in cat_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
```

```python
#STEP 7.3: Verify No Missing Values Remain
df_clean.isnull().sum().sum()
```

```python
#STEP 8: Remove Duplicate Rows
#STEP 8.1: Check Duplicate Rows
df_clean.duplicated().sum()
```

```python
#STEP 8.2: Remove Duplicate
df_clean.drop_duplicates(inplace=True)
```

```python
#STEP 8.3: Verify Shape After Removing Duplicates
df_clean.shape
```

```python
#STEP 9: Remove Invalid / Physically Impossible Values
#STEP 9.1: Apply Validity Rules
df_clean = df_clean[df_clean['pl_rade'] > 0]      # Planet radius must be positive
df_clean = df_clean[df_clean['pl_orbper'] > 0]   # Orbital period must be positive
df_clean = df_clean[df_clean['pl_eqt'] > 0]      # Temperature must be positive
```

```python
#STEP 9.2: Check Shape After Validation
df_clean.shape
```

```python
#STEP 10: Final Validation & Save Output
#STEP 10.1: Final Validation (Summary Statistics)
df_clean.describe()
```

```python
df_clean.to_csv("../data/processed/preprocessed.csv", index=False)
print("Phase 1 preprocessing completed successfully.")
```
